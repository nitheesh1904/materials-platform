```python
from pathlib import Path
import json

repo_root = Path.cwd().parent

run_dir = repo_root / "lammps" / "Si" / "runs"

# Get the job directories
run_dirs = [p for p in run_dir.iterdir() if p.is_dir()]

# Get latest job
latest_run = max(run_dirs, key=lambda p: p.stat().st_mtime)

# Read job.json
with open(latest_run / "run.json") as f:
    job = json.load(f)

job_id = job["run_id"]

print(job_id)
```
